# Proposed Model 5 — Universal Target Memory
SAM2 + Grounding DINO + SigLIP 2 + trusted motion/geometry/color memory + geometry validation + recovery gating.
**Use a Colab GPU runtime.** Camera-motion estimates are diagnostic, not a reliable world-coordinate tracker.


```python
from pathlib import Path
import sys,subprocess,urllib.request,torch
repo=Path('/content/Target_Locker')
branch='feature/proposed-model-5-universal-memory'
if not torch.cuda.is_available(): raise RuntimeError('Colab GPU required')
if not repo.exists(): subprocess.run(['git','clone','https://github.com/CptImtiaz/Target_Locker.git',str(repo)],check=True)
subprocess.run(['git','-C',str(repo),'fetch','origin',branch],check=True)
subprocess.run(['git','-C',str(repo),'checkout','-B',branch,'FETCH_HEAD'],check=True)
p=repo/'proposed_model_5'
subprocess.run([sys.executable,'-m','pip','install','-q','-r',str(p/'requirements.txt')],check=True)
ckpt=Path('/content/sam2.1_hiera_tiny.pt')
if not ckpt.exists(): urllib.request.urlretrieve('https://dl.fbaipublicfiles.com/segment_anything_2/092824/sam2.1_hiera_tiny.pt',ckpt)
print('Model 5 code installed:',p)
```

```python
import subprocess,sys
subprocess.run([sys.executable,'-m','unittest','discover','-s',str(p/'tests'),'-v'],check=True)
subprocess.run([sys.executable,'-m','compileall','-q',str(p)],check=True)
print('CPU tests and syntax checks completed')
```
